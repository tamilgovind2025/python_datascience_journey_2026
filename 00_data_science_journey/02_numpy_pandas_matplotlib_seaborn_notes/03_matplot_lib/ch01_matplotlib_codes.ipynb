# Sorting Bars

For ranking problems, sorting is often useful.

```python
import matplotlib.pyplot as plt 

products = ["Laptop", "Mobile", "Tablet", "Watch"]
sales = [120, 250, 100, 180]
```

```python
zip_data=zip(sales,products)

print(list(zip_data))
```

```python
sorted_data=sorted(zip(sales,products),reverse=True)
print(sorted_data)
```

```python
zip_data=zip(sales,products)
sorted_data=sorted(zip_data)
print(sorted_data)
```

```python
sales_sorted,products_sorted=zip(*sorted_data)
plt.barh(products_sorted,sales_sorted)
plt.title("products vs sales sorted")
plt.xlabel("sales")
plt.ylabel("products")
plt.show()
```

# complete examples of sorted data 

```python
sales = [5000, 12000, 8000, 15000]
products = ["Laptop", "Mobile", "Tablet", "Monitor"]

sorted_data = sorted(
    zip(sales, products),
    reverse=True
)

sales_sorted, products_sorted = zip(*sorted_data)

print(sales_sorted)
print(products_sorted)
```

```python
import matplotlib.pyplot as plt 

sales = [5000, 12000, 8000, 15000]
products = ["Laptop", "Mobile", "Tablet", "Monitor"]

sorted_data = sorted(
    zip(sales, products),
    reverse=True
)

sales_sorted, products_sorted = zip(*sorted_data)

plt.barh(products_sorted, sales_sorted)

plt.xlabel("Sales")
plt.title("Product Sales Ranking")

plt.show()
```

```python
#multiple bar graph 
import matplotlib.pyplot as plt
import numpy as np

products = ["Laptop", "Mobile", "Tablet"]

sales_2025 = [50, 80, 40]
sales_2026 = [70, 90, 60]

x=np.arange(len(products))
width=0.35

plt.bar(x-width/2,sales_2025,width,label="sales_2025")
plt.bar(x+width/2,sales_2026,width,label="sales_2026")
plt.xticks(x,products)
plt.xlabel("products")
plt.ylabel("sales")
plt.title("product comparision 25 vs 26")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

marks = np.random.randint(0, 101, 100)

plt.figure(figsize=(8, 5))

counts, bins, patches = plt.hist(
    marks,
    bins=10,
    edgecolor="black",
    alpha=0.8
)

# Display count on top of each bar
for count, patch in zip(counts, patches):
    plt.text(
        patch.get_x() + patch.get_width() / 2,
        count,
        int(count),
        ha="center",
        va="bottom",
        fontsize=10,
        fontweight="bold"
    )

plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.grid(axis="y", alpha=0.4)

plt.tight_layout()
plt.show()
```

```python
#symmetric distribution 
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

marks = np.random.normal(
    loc=60,
    scale=10,
    size=1000
)

plt.hist(marks, bins=20, edgecolor="black")

plt.title("Student Marks - Symmetric Distribution")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()
```

```python
#right skew 
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

marks = np.random.lognormal(
    mean=3.8,
    sigma=0.35,
    size=1000
)

plt.hist(marks, bins=20, edgecolor="black")

plt.title("Student Marks - Right Skewed")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

marks = 100 - np.random.lognormal(
    mean=3.8,
    sigma=0.35,
    size=1000
)

plt.hist(marks, bins=20, edgecolor="black")

plt.title("Student Marks - Left Skewed")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# 1. Symmetric
symmetric = np.random.normal(60, 10, 1000)

# 2. Right skewed
right_skew = np.random.lognormal(3.8, 0.35, 1000)

# 3. Left skewed
left_skew = 100 - np.random.lognormal(3.8, 0.35, 1000)

# Symmetric
plt.figure(figsize=(7, 4))
plt.hist(symmetric, bins=20, edgecolor="black")
plt.title("Symmetric Distribution")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()

# Right skew
plt.figure(figsize=(7, 4))
plt.hist(right_skew, bins=20, edgecolor="black")
plt.title("Right-Skewed Distribution")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()

# Left skew
plt.figure(figsize=(7, 4))
plt.hist(left_skew, bins=20, edgecolor="black")
plt.title("Left-Skewed Distribution")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import skew

np.random.seed(42)

marks = np.random.normal(60, 10, 1000)

mean = np.mean(marks)
median = np.median(marks)
skewness = skew(marks)

plt.figure(figsize=(8, 5))

plt.hist(
    marks,
    bins="auto",
    edgecolor="black",
    alpha=0.8
)

plt.axvline(
    mean,
    linestyle="--",
    linewidth=2,
    label=f"Mean = {mean:.2f}"
)

plt.axvline(
    median,
    linestyle="-",
    linewidth=2,
    label=f"Median = {median:.2f}"
)

plt.title("Student Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Frequency")

plt.legend()
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()

print("Mean:", mean)
print("Median:", median)
print("Skewness:", skewness)
```
