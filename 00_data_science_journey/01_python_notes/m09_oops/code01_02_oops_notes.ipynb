# class attributes

```python
#class attributes (datas)
class student:
    name="ram"
    age=25
    dept="ECE"
    
#how to access 
#2 ways 
#1st way getattr(class_name,attributes,optional_ifnot avalible this will print)

print(getattr(student,"name"))
print(getattr(student,"age"))
print(getattr(student,"dept"))
print(getattr(student,"marks","not available"))
print(getattr(student,"marks")) #it will throw error 
```

```python
# 2 . notation method mostly used 
print(student.name)
print(student.age)
print(student.dept)
print(student.marks) #this will throw attribute error
```

# instance attributes

```python
class employee():
    company_name="hcl tech"
    branch="chennai"

obj1=employee()
print(employee.__dict__) #class name space
print(obj1.__dict__) #object namespace
```

```python
print(obj1.company_name)
```

```python
print(obj1.branch)
```

```python
obj2=employee()
obj2.branch="bangalore"
print(obj2.branch)

obj3=employee()
print(obj3.branch)
```

```python
#class atrributes(data) and behavior (methods)
class student:
    name="govind"
    age=21
    dept="ECE"
    
    def print_all():
        print("name = ",student.name)
        print("age = ",student.age)
        print("dept = ",student.dept)
```

```python
student.print_all() #class methods
```

```python
print(getattr(student,"print_all"))
```

```python
getattr(student,"print_all")()
```

```python
print(student.__dict__)
print(student.__dict__["print_all"])
```

```python
student.__dict__["print_all"]()
```

# instance methods

```python
class student:
    name="tamil govind"
    age=25
    
    def print_all(self):
        print("name = ",student.name)
        print("age =",student.age)
        
```

```python
o=student()
```

```python
o.print_all()
```

```python
student.print_all()
```

```python
student.print_all(o)
o.print_all()
```

```python
class student:
    name="govind"
    age=36
    dept="data science"
    
    def print_all(self,gender):
        print("name = ",student.name)
        print("age = ",student.age)
        print("dept = ",student.dept)
        print("gender = ",gender)
```

```python
o=student()
o.print_all("male")
```

# constructor __init__method

```python
class user:
    def __init__(self):
        print("init method executed")
obj1=user()
obj2=user() #automatically executed when new instance created
```

```python
class student:
    pass
#obj1=student("name") #we cannot assign values here 
obj1=student()
obj1.name="govind"
print(obj1.__dict__)
print(student.__dict__)
```

```python
class user:
    def __init__(self,name,age):
        print("init started")
        self.name=name
        self.age=age
        print("init completed")
        
    def print_all(self):
        print("name = ",self.name)
        print("name = ",self.age)
        
obj1=user("govind",35)
obj1.print_all()
```

```python
obj2=user("vinoth",25)
obj2.print_all()
```
