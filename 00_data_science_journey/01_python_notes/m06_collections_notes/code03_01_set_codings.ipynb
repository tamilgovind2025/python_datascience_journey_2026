```python
numbers = {10,20,30,40,50,60}

result = set(filter(lambda x: x > 30, numbers))

print(result)

#map modify 
#filter select 
```

```python
#example one more filters 
numbers = {10,11,12,13,14,15}

even = set(filter(lambda x: x % 2 == 0, numbers))

print(even)
```

```python
result = set(
    map(
        lambda x: x**2,
        filter(lambda x: x % 2 == 0, numbers)
    )
)

print(result)
```
