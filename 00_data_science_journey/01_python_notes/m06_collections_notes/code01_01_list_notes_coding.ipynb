# list comprehension 

```python
numbers=list(range(1,11))
print(numbers)

#square the list values 
square=[]
for i in numbers:
    square.append(i**2)
print(square)

#list comprehension 
square=[x**2 for x in numbers] #this is list comprehension
print(square)
```

```python
#list comprehension with condition 
numbers=list(range(1,21))
print(numbers)

even=[x for x in numbers if x%2==0]
print(even)
odd=[x for x in numbers if x%2==1]
print(odd)
```

```python
#print values from 1 to 100 which is divisile by 4 
numbers=list(range(1,101))
print(numbers)

data=[x for x in numbers if not x%4]
print(data)

data=[x for x in numbers if x%4==0]
print(data)
```

```python
#conditional expression 
data =[98,86,45,32,78,43,23,89]

#general 
result=[]
for mark in data:
    if mark>=35:
        result.append("Pass")
    else:
        result.append("Fail")
print(result)

#conditional expression 
result =["Pass" if mark>=35 else "Fail" for mark in data]
print(result)
```

```python
#nested list comprehension
matrix =[
    [1,2,3],
    [4,5,6]
]
print(matrix)

#general 
for row in matrix:
    for value in row:
        print(value)
        
#nested comprehension
flatten=[value for row in matrix for value in row]
print(flatten)
```

```python
#list copy vs shallow copy vs deep copy 
a=[22,24,65,76,32,45]

b=a #alias name for a because both reference same object 
print(id(a))
print(id(b)) #if b changes a also changes 

#shallow copy  #copied only outer object not inner object best example is nested lists 
c=a.copy()
print(id(c))
a[1]=100
print(a)
print(b)
print(c)
```

```python
#nested copy problem 
data=[
    [22,33,44],
    [44,55,66]
]

#shallow copy issue 
shallow_data=data.copy()

print(id(data))
print(id(shallow_data))

#now both refers same object 
shallow_data[0][2]=999
print(shallow_data)
print(data)
```

```python
print(id(data[0]))
print(id(shallow_data[0]))
```

```python
#to over come this issue deepcopy we have to use 
#note this all methods are shallow copy 
a=[1,2,3]
b=a.copy() 
print(b)
b=list(a)
print(b)
c=a[:]
print(c)
```

```python
import copy 
a=[1,2,3,4,5]
b=copy.deepcopy(a)  #now totally new object including members 
print(b)
```
