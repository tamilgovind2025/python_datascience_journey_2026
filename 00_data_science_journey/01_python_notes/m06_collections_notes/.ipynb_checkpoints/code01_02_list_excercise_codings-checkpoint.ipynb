```python
numbers = [10, 20, 10, 30, 20, 10]
print("number of elements = ",len(numbers))
print("number of 10's = ",numbers.count(10))
print("index of first 30 = ",numbers.index(30))
print("unique values = ",list(set(numbers)))
print("maximum value = ",max(numbers))
print("minimum value = ",min(numbers))
print("total value  = ",sum(numbers))

#set is unorderd, if order matters use this method
unique=list(dict.fromkeys(numbers))
print(unique)

#or we can write the code 
unique=[]
for i in numbers:
    if i not in unique:
        unique.append(i)
print(unique)
```

```python
marks = [35, 67, 89, 45, 92, 28, 76, 55]
print("average mark = ",sum(marks)/len(marks))
print("highest mark = ",max(marks))
print("lowest mark  = ",min(marks))
passed_students=[x for x in marks if x>=35]
print("passed students = ",passed_students)
failed_students=[x for x in marks if x<35]
print("failed students = ",failed_students)
print("passed students count = ",len(passed_students))
print("failed students count = ",len(failed_students))
print("everybody passed = ",all(mark>=35 for mark in marks))
print("any one scored above 90 = ",any(mark>90 for mark in marks))
```

```python
employees = [
    {"name": "Arun", "salary": 45000, "department": "IT"},
    {"name": "Bala", "salary": 65000, "department": "HR"},
    {"name": "Chitra", "salary": 85000, "department": "IT"},
    {"name": "Deepak", "salary": 55000, "department": "Finance"},
    {"name": "Ezhil", "salary": 95000, "department": "IT"}
]


# 1. Extract names
names = [employee["name"] for employee in employees]
print("Names:", names)


# 2. Extract salaries
salaries = [employee["salary"] for employee in employees]
print("Salaries:", salaries)


# 3. Highest-paid employee
highest_paid = max(
    employees,
    key=lambda employee: employee["salary"]
)
print("Highest paid:", highest_paid)


# 4. Lowest-paid employee
lowest_paid = min(
    employees,
    key=lambda employee: employee["salary"]
)
print("Lowest paid:", lowest_paid)


# 5. Employees earning > 60000
high_salary = [
    employee
    for employee in employees
    if employee["salary"] > 60000
]
print("Salary > 60000:", high_salary)


# 6. IT employees
it_employees = [
    employee
    for employee in employees
    if employee["department"] == "IT"
]
print("IT employees:", it_employees)


# 7. Sort by salary
ascending = sorted(
    employees,
    key=lambda employee: employee["salary"]
)
print("Ascending:", ascending)


# 8. Sort by salary descending
descending = sorted(
    employees,
    key=lambda employee: employee["salary"],
    reverse=True
)
print("Descending:", descending)


# 9. Average salary
average_salary = sum(
    employee["salary"] for employee in employees
) / len(employees)

print("Average salary:", average_salary)


# 10. Anyone earning > 1 lakh?
result = any(
    employee["salary"] > 100000
    for employee in employees
)

print("Anyone earning > 1 lakh?", result)
```
