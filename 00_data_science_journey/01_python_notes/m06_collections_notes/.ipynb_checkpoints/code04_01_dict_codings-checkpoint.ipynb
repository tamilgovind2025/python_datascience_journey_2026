```python
#['clear', 'copy', 'fromkeys', 'get', 'pop', 'popitem', 'setdefault', 'update', ]
#'keys','values','items'

data={
    "name":["govind","kumar","vinoth"],
    "age":[36,35,23]
}
print(data)

#keys()
print(data.keys())
#values()
print(data.values())
#items
print(data.items()) #return dictionary view objects:
```

```python
#[ 'fromkeys', 'pop', 'popitem', 'setdefault', 'update', ]
#'get','copy','clear',

data={
    "name":["govind","kumar","vinoth"],
    "age":[36,35,23]
}
print(data)
#get
print(data.get("name"))
print(data.get("marks",{}))
#copy - shallow copy
data_new=data.copy()
print(id(data))
print(id(data_new))
print(id(data["name"]))
print(id(data_new["name"]))

```

```python
print(data_new["name"])
```

```python
data_new["name"][0]="tamil govind"
print(data_new)
print(data)
```

```python
#to overcome this issue use deep copy 
import copy 
data={
    "name":["govind","kumar","vinoth"],
    "age":[36,35,23]
}
print(data)

data_new=copy.deepcopy(data)
print(data_new)

print(id(data_new))
print(id(data))

print(id(data_new["name"]))
print(id(data["name"]))

data_new["name"][0]="vinoth kumar"
print(data_new)
print(data)
```

```python
print(data_new)
data_new.clear()
print(data_new)
```

```python
#[ 'fromkeys',  'setdefault' ]
#'pop', 'popitem',
print(data)
print(data.pop("name"))
print(data)
```

```python
#[ 'fromkeys',  'setdefault' ]
#Real-time use
#Suppose you have student names and initially want to assign all of them a status:
students = ["Govind", "Kumar", "Vinoth"]
status = dict.fromkeys(students, "Pending")
print(status)
```

```python
data={
    "name":"govind",
    "age":35,
}
print(data)
print(data.setdefault("name","vinoth kumar"))
print(data)
print(data.setdefault("city","chennai"))
print(data)

#Here "city" doesn't exist.
#Therefore Python:
#Creates "city"
#Assigns "Chennai"
#Returns "Chennai"

#normal assigment  overwrites 
data["city"] = "bangalore"
print(data)

#but setdefault if exist will not disturb
print(data.setdefault("city","chennai"))
print(data)
```

```python
#max()
data={
    "student_names":["vinoth kumar","govind","gowtham","surya","vimal","jothi"],
    "marks":[98,97,56,78,89,88,76]
}
print(data)

print(max(data)) #default is key it returns max key length
print(max(data["student_names"]))
#print(max(data.values())) #here it contains string and int so cannot compare
print(max(data["marks"]))
```

```python
#min()
data={
    "student_names":["vinoth kumar","govindarasan","gowtham","surya","vimal","jothi"],
    "marks":[98,97,56,78,89,88]
}
print(data)
print(data.keys())
print(min(data)) #min by default it works keys 
print(min(data["student_names"])) #String comparison is based on Unicode/lexicographical order.i mean char by char
print(min(data["marks"]))
```

```python
#min() works with two lists 
data={
    "tamil":[78,97,78,96,76,65],
    "english":[98,98,67,87,86,97],
    "maths":[99,96,45,32,89,92]
}
print(data)

print(min(data))
print(min(data.values()))
print(min(data["tamil"]))
print(min(data["english"]))
```

```python
#sum()
data={
    "student_names":["vinoth kumar","govindarasan","gowtham","surya","vimal","jothi"],
    "marks":[98,97,56,78,89,88]
}
print(data)
help(sum)
```

```python
data={
    "student_names":["vinoth kumar","govindarasan","gowtham","surya","vimal","jothi"],
    "marks":[98,97,56,78,89,88]
}
print(data)
#print(sum(data)) cannot work
#print(sum(data["student_names"])) can not work
print(sum(data["marks"])) #it mainly work with numbers only 
```

```python
#all() and any()
data={
    "student_names":["vinoth kumar","govindarasan","gowtham","surya","vimal","jothi"],
    "marks":[98,97,56,78,89,88],
    "average":[23,54,76,0,0,0],
}
print(data)
print(all(data))
print(all(data["student_names"]))
print(all(data["average"]))

#any()
print(any(data))
print(any(data["student_names"]))
print(any(data["average"]))
```

```python
#len()
data={
    "student_names":["vinoth kumar","govindarasan","gowtham","surya","vimal","jothi"],
    "marks":[98,97,56,78,89,88],
    "average":[23,54,76,0,0,0],
}
print(data)
print(len(data))

print(len(data["student_names"]))
print(len(data.values()))
print(len(data.items()))
```

# inbuilt methods


```python
#sum() it works with numbers only 
marks = {
    "Govind": 95,
    "Kumar": 88,
    "Vinoth": 76,
    "Surya": 92
}

#print(sum(marks)) it will throw error 
print(sum(marks.values()))

#average marks 
print("average  marks = ",sum(marks.values())/len(marks))
```

```python
#Monthly expenses
expenses = {
    "Rent": 12000,
    "Food": 8000,
    "Travel": 4000,
    "Electricity": 2500,
    "Internet": 1000
}

#monthly expenses 
total=sum(expenses.values())
print(total)
```

```python

#max() finds the largest item.

marks = {
    "Govind": 95,
    "Kumar": 88,
    "Vinoth": 76,
    "Surya": 92
}

print(max(marks)) #it will return key by default

#value 
print(max(marks.values()))
name,mark=max(marks.items(),key=lambda x: x[1])
print("the max marks is",mark,"and scored by",name)

#find the highest mark 
print(max(marks,key=marks.get)) find the maximum key based on its value
```

```python
salary = {
    "Arun": 45000,
    "Bala": 60000,
    "Kumar": 55000,
    "Ravi": 75000
}

#highest salary 
print(max(salary,key=salary.get)) #dont give get() that is fucntion execute but max call that object 
```

```python
#min() — Minimum
marks = {
    "Govind": 95,
    "Kumar": 88,
    "Vinoth": 76,
    "Surya": 92
}

lowyest=min(marks,key=marks.get)
print(lowyest)
print(marks[lowyest])

#key 
print(min(marks))
#values
print(min(marks.values()))
#key associate with value 
print(min(marks,key=marks.get))
```

```python
#all() - check all values true it return boolean True or False 
marks = [80, 90, 75, 88]

result=all(mark>=50 for mark in marks)
print(result)
```

```python
#check attendacne percetage above 75%

attendance = {
    "Arun": 85,
    "Bala": 92,
    "Kumar": 78,
    "Ravi": 65
}

result=all(data>=75 for data in attendance.values())
print(result)
```

```python
#any() atleast one true 
marks = [35, 45, 78, 90]
result=any(mark>=90 for mark in marks)
print(result)
```

```python
#len() count 
students = {
    "Arun": 85,
    "Bala": 92,
    "Kumar": 78,
    "Ravi": 95
}

print(len(students))
```

```python
#enumerate()

students = {
    "Arun": 85,
    "Bala": 92,
    "Kumar": 78,
    "Ravi": 95
}

#keys only 
for index,data in enumerate(students):
    print(index,data)
#values only 
for index,data in enumerate(students.values()):
    print(index,data)
    
#key values pairs as tuple 
#values only 
for index,data in enumerate(students.items()):
    print(index,data)
    
#extract tuple  best way is this method. remember enumeratde doesnot change the dict. just give counter while iterate
for index,(name,mark) in enumerate(students.items()):
    print(index,name,mark)
```

```python
#zip() multiples items combines together 

names = ["Govind", "Kumar", "Vinoth"]
marks = [95, 88, 76]

data=dict(zip(names,marks))
print(data)
```

```python
#multiple dict 
names = ["Govind", "Kumar", "Vinoth"]
marks = [95, 88, 76]
attendance = [92, 85, 78]

data=dict(zip(names,zip(attendance,marks)))
print(data)
```

```python
#map() means: Apply a function to every element.

marks = {
    "Govind": 80,
    "Kumar": 70,
    "Vinoth": 90
}

#add extra 5 marks for all students 

#print(marks.items())
new_marks=dict(map(lambda x:(x[0],x[1]+5),marks.items()))
print(new_marks)
```

```python
#salary increment calculation 10% increment

salary = {
    "Arun": 30000,
    "Bala": 40000,
    "Kumar": 50000
}

new_salary=dict(map(lambda x:(x[0],x[1]*1.10),salary.items()))
print(new_salary)
```

```python
#filter() functions 
marks = {
    "Govind": 95,
    "Kumar": 45,
    "Vinoth": 78,
    "Surya": 32
}

filter_data=dict(filter(lambda x: x[1]>=45,marks.items()))
print(filter_data)
```

```python
#eligible students for exam 
#attendance percentage should be above 75%

attendance = {
    "Arun": 85,
    "Bala": 62,
    "Kumar": 48,
    "Ravi": 91
}

filter_data=dict(filter(lambda x: x[1]>=75,attendance.items()))

#function 
def check_attendace(x):
    return x[1]>=75

filter_data=dict(filter(check_attendace,attendance.items()))
filter_data
```

```python
#range() used in dict 

students = {
    "Arun": 85,
    "Bala": 92,
    "Kumar": 78
}

key=list(students.keys())

for i in range(len(key)):
    print(i,key[i],students[key[i]])
```

```python
#reversed()

students = {
    "Arun": 85,
    "Bala": 92,
    "Kumar": 78,
    "Ravi": 95
}

data=list(reversed(students))  #reversed always returns list 
print(data)
```
