# Real-World Applications

```python
username1="govind_123"
username2="admin@123"

valid=(5<len(username1)<15 and all(x.isalnum() or x=="_" for x in username1))
print(valid)

#tradional way 
data=[]
for x in username1:
    if x.isalnum() or x=="_":
        data.append(True)
    else:
        data.append(False)

if 5<len(username1)<15 and all(data):
    print(True)
else:
    print(False)
```

```python
password = "Govind2@23"
has_length=len(password)
has_upper=any(x.isupper() for x in password)
has_lower=any(x.islower() for x in password)
has_number=any(x.isdigit() for x in password)
has_special=any(not x.isalnum() for x in password)
if has_length and has_upper and has_lower and has_number and has_special:
    print("valid")
else:
    print("invalid")
    
#another way 
valid=all([has_length,has_upper,has_lower,has_number,has_special])
print(valid)
```

```python
text="python is a powerful programming language in the world"
words=text.split()
max_word=max(words,key=len)
print(max_word)
print(len(max_word))

for i in words:
    print(len(i))
```

```python
sentence = "Python programming language"

#1 finding longest word
words=sentence.split(" ")
print(words)
word=max(words,key=len)
print(word)
print(max(sentence.split(" "),key=len))

#2length of longest word
print(len(word))
print(len(max(sentence.split(" "),key=len)))

#3. Sum of Unicode values
sum_max_word=sum(map(ord,word))
print(sum_max_word)
```

```python
text = "Python Programming 2026"

#1. Total number of characters
print("total number of chars = ",len(text))

#2. Number of alphabetic characters
print("total num of alphabatic chars = ",sum(x.isalpha() for x in text))

#3. Number of digits
print("total num of digits = ",sum(x.isdigit() for x in text))

#4. Number of spaces
print("total number of spaces = ",sum(x.isspace() for x in text))

#5. Highest Unicode character
print("highest unicode char = ",max(set(text.replace(" ","")),key=ord))

#6. Lowest Unicode character
print("lowest unicode char = ",min(set(text.replace(" ","")),key=ord))

#7. Sum of Unicode values
print("sum of unicode values = ",sum(map(ord,text)))

#8. Whether all characters are ASCII
print("are all characters ascii = ",all(ord(x)<128 for x in text))
```
