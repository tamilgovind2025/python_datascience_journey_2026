```python
#max(),min(),sorted(),reveresed(),zip(),enumerate(),map(),filter(),
a=(22,34,56,78,95,22,34,34,22)
print("maximum value = ",max(a))
print("minimum value = ",min(a))

#sorted returns value 
sorted_data=sorted(a) #by default ascending order want to change descending sorted(a,reverse=True)
print(sorted_data)
sorted_data=sorted(a,reverse=True)
print(sorted_data)
```

```python
#zip(),

names=("govind","vinoth","kumar","kavitha") #it will return matching values only shorted values omitted donot give error 
marks=(98,78,96,67)

#zip 
new_data=tuple(zip(names,marks))
print(new_data)
```

# tuple packing and unpacking 

```python
student_data="govind",22,"ECE",98.5,3
print(student_data) #python automatically packint to tuple 

#unpacking 
name,age,dept,attend,year=student_data
print(name)
print(year)
```

# tuple hashability 

```python
#a tuple can be used as dict key if all the elements are hashable 
#ex hashable 

location={
    (10,20):"chennai",
    (10,22):"bangalore"
}

print(location)
print(location[(10,20)])
```

```python
#hashability checking 
a=((1,2),(2,4))
print(hash(a))

a=([1,2])
print(hash(a))
```

```python
#hashable tuple 
#golden rule - a tuple is hashable only if all the elements are hashable 
#tuple can be used as dict key 
t=(10,"python",3.5)
print(hash(t))
```

```python
#tuple as function return value 

def calculate(a,b):
    return a+b,a-b,a*b,a/b

result=calculate(10,20)
print(result)
```

```python
#realtime example 

def analyze_marks(marks):
    return (max(marks),min(marks),sum(marks),sum(marks)/len(marks))

values=(20,30,40)
result=analyze_marks(values)
print(result)
```
