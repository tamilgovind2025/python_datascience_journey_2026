```python
# [ID, Name, Department, Year, Attendance, Math, Python, DBMS, OS, CN, ML]
departments=set(data[2] for data in students)

for dept in departments:
    dept_students=[data for data in students if data[2]==dept]
    lowyest_st=min(dept_students,key=lambda x:x[4])
    print(dept,lowyest_st[1],lowyest_st[4])
```

```python
# [ID, Name, Department, Year, Attendance, Math, Python, DBMS, OS, CN, ML]
departments=set(data[2] for data in students)

for dept in departments:
    dept_students=[data for data in students if data[2]==dept]
    print("dept = ",dept,"student count = ",len(dept_students))
```

```python
# [ID, Name, Department, Year, Attendance, Math, Python, DBMS, OS, CN, ML]

for data in students:
    if any(mark<50 for mark in data[5:]):
        print(data[1])
```

```python
departments=set(data[2] for data in students)
for dept in departments:
    dept_students=[data for data in students if data[2]==dept]
    dept_marks=sum(sum(marks[5:])for marks in dept_students)
    length=len(dept_students)*len(dept_students[0][5:])
    print(dept,dept_marks,round(dept_marks/length,2))
```

```python
ranked_students=sorted(students,key=lambda data: sum(data[5:])/len(data[5:]),reverse=True)
#print(ranked_students[:3])
for i in range(3):
    print(ranked_students[i])
```

```python
# [ID, Name, Department, Year, Attendance, Math, Python, DBMS, OS, CN, ML]

subjects={
    "Math":5,
    "Python":6,
    "DBMS":7,
    "OS":8,
    "CN":9,
    "ML":10
}

for subject,index in subjects.items():
    highest_mark=max(student[index] for student in students)
    toppers=[student[1] for student in students if student[index]==highest_mark]
    print(subject,":","highest mark =",highest_mark,"->",toppers)
```
