# string conversion 

```python
#remember any kind of data can be easily converted to string because. string can handles letters,numbers,special chars 

#int to string
a=10
b=str(a)
print(b)
print(type(b))

#float to string
a=10.5
b=str(a)
print(b)
print(type(b))

#bool to string
a=True
b=str(a)
print(b)
print(type(b))

#complex to string
a=3+3j
b=str(a)
print(b)
print(type(b))
a=complex(b)
print(a)
print(type(a))
```

```python
#list to string 
list_data=[12,13,15]

str_data=str(list_data)
print(str_data)
print(type(str_data))

#role back  #connnot do ti for collections to overcome this 
data=list(str_data)
print(data)

#some other way   why we need this 
#in real time data comes as string. but they will give as string list so difficulty to handle to over come this issue 
#use this method

# it is extremely important when handling input

import ast  #abstract syntax tree
data=ast.literal_eval(str_data) #interpret this text as python literal
print(data)
print(type(data))
```

```python
#Group 3-Searching Inside a String
#find(),rfind(),index(),count(),rindex()

text = "welcome to python programming"
#help(str.find)
s1="python"
s2="java"
print(text.find(s1)) #return index position if not found return -1. 
print(text.find("o"))
print(text.find("o",5))
print(text.find("o",10,20))
print(text.find("ml",10,50)) #stop excluded and if
```

```python
#translate() is a string method used to replace multiple characters at once.
#It is especially useful when you want to replace characters based on a mapping.
#help(str.translate)
data="i love python computer programming"
table=str.maketrans("aeiou","12345")
print(data.translate(table))
```

```python
#Remove punctuation from user input:
#str.maketrans(from, to, delete)
#maketrans()- Create translation table

text = "Hello, Govind! How are you?"
table=str.maketrans("","","?!,")
print(text.translate(table))
```

```python
#Cleaning phone numbers
numbers = [
    "+91-98765-43210",
    "+91 98765 43210",
    "(91) 98765-43210"
]

table=str.maketrans("","","+-() ")
cleaned_data=[]
for num in numbers:
    cleaned_data.append(num.translate(table))
cleaned_data

    
```

```python
#saalry calculation 
salary=["₹50,000","₹75,000","₹1,00,000"]

table=str.maketrans(",","_","₹")
salary_data=[]
for data in salary:
    salary_data.append(int(data.translate(table)))
    
salary_data
```
