# binary file copy and paste

```python
#image.jpg copy and  write as copy_image.jpg 

#read the data
with open(r"dataset\image.jpg","rb") as file:
    data=file.read()
#copy the data 
with open(r"dataset\copy_image.jpg","wb") as file:
    file.write(data)
    print("copy and paste done ")
```

# copy and paste - production style chunk by chunk

```python
#mainly used for large file systems to handle efficiently 

with open(r"dataset\image.jpg","rb") as source:
    with open(r"dataset\chunk_copy_image.jpg","wb") as dest:
        while True:
            chunk=source.read(1024) #1kb data read 
            if not chunk:
                break
            dest.write(chunk)
print("data copy and paste work done ")
```
