# Level 1 — Student Management System

Create a program to manage student information.

```python
students = [
    {
        "id": 101,
        "name": "Arun",
        "department": "CSE",
        "marks": [85, 78, 92, 88, 76],
        "skills": {"Python", "SQL", "Git"}
    },
    {
        "id": 102,
        "name": "Priya",
        "department": "ECE",
        "marks": [90, 82, 88, 91, 95],
        "skills": {"Python", "Java", "SQL"}
    },
    {
        "id": 103,
        "name": "Karthik",
        "department": "CSE",
        "marks": [65, 72, 68, 70, 75],
        "skills": {"Python", "HTML"}
    }
]
```

```python
#01.Display all student names.
for data in students:
    print(data["name"])
```

```python
#Display students belonging to CSE.
for data in students:
    if data["department"]=="CSE":
        print(data)
        
#list comprehension 
cse_students=[data for data in students if data["department"]=="CSE"]
print(cse_students)
```

```python
#Calculate each student's total marks.
for data in students:
    print(data["name"],sum(data['marks']))
```

```python
#Calculate average marks.
for data in students:
    print(data["name"],sum(data['marks'])/len(data["marks"]))
```

```python
#Find the topper.
#help(max)

max_mark=max(students,key=lambda x:sum(x["marks"]))
print(max_mark)
print(max_mark["name"])
```

```python
#Find students with average > 80.
for data in students:
    if sum(data["marks"])/len(data["marks"])>80:
        print(data["name"])
```

```python
#Find students who failed in any subject.
for data in students:
    for mark in data["marks"]:
        if mark<=40:
            print(data["name"])
            break
            
#alternative way           
for student in students:
    if any(mark <= 40 for mark in student["marks"]):
        print(student["name"])
```

```python
#Find all unique skills.
all_skills=set()

for data in students:
    all_skills=all_skills.union(data["skills"])
    
print(all_skills)
```

```python
#Find students who know Python.
for data in students:
    if "Python" in data["skills"]:
        print(data["name"])
```

```python
#Find students who know both Python and SQL.

for data in students:
    if "Python" in data["skills"] and "SQL" in data["skills"]:
        print(data["name"])
```

```python
#Find the student with the highest number of skills.

student=max(students,key=lambda x: len(x["skills"]))
print(student["name"])
```

```python
#Create a function:
def calculate_average(marks):
    return sum(marks)/len(marks)

for data in students:
    avg_mark=calculate_average(data["marks"])
    print(data["name"],avg_mark)
```

```python
def get_grade(average):
    if average>=90:
        return "S"
    elif 85<=average<90:
        return "A+"
    elif 80<=average<85:
        return "A"
    elif 75<=average<80:
        return "B+"
    elif 70<=average<75:
        return "B"
    else:
        return "Fail"

print(f'{"ID":<10}{"Name":<10}{"Dept":<10}{"Average":<10}{"Grade":<10}')
for data in students:
    avg_mark=calculate_average(data["marks"])
    grade=get_grade(avg_mark)
    print(f'{data["id"]:<10}{data["name"]:<10}{data["department"]:<10}{avg_mark:<10}{grade:<10}')
```

# Level 2 — Employee Salary System

This is closer to a company application.

```python
employees = [
    (101, "Arun Kumar", "Python Developer", 60000, "Chennai"),
    (102, "Priya", "Data Analyst", 55000, "Bangalore"),
    (103, "Rahul", "ML Engineer", 85000, "Chennai"),
    (104, "Divya", "Python Developer", 70000, "Hyderabad"),
    (105, "Karthik", "Data Analyst", 50000, "Chennai")
]
```

```python
#(101, "Arun Kumar", "Python Developer", 60000, "Chennai")
#Display all employees.
for data in employees:
    print(data[0],data[1])
    
#tuple unpacking 
for empid,name,role,salary,location in employees:
    print(empid,name)
```

```python
#(101, "Arun Kumar", "Python Developer", 60000, "Chennai"),
#Display employees working in Chennai.
for empid,name,role,salary,location in employees:
    if location=="Chennai":
        print(empid,name)
        
#Display employees working in Chennai.
for empid,name,_,_,location in employees:
    if location=="Chennai":
        print(empid,name)
```

```python
#(101, "Arun Kumar", "Python Developer", 60000, "Chennai"),
#Find highest salary.
_,name,_,highest_salary,_=max(employees,key=lambda x: x[3])
print(name,highest_salary)
```

```python
#(101, "Arun Kumar", "Python Developer", 60000, "Chennai"),
#Find lowest salary.
_,name,_,lowest_salary,_=min(employees,key=lambda x: x[3])
print(name,lowest_salary)
```

```python
#(101, "Arun Kumar", "Python Developer", 60000, "Chennai"),
#Calculate average salary.
average_salary=sum(employee[3] for employee in employees)/len(employees)
print(average_salary)
```
