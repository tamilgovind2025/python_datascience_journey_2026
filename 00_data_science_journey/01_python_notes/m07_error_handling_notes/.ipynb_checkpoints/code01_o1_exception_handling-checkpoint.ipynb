# multiple exception handling 

```python
try:
    x=int(input("enter the number "))
    result=100/x
    print(result)
except ValueError:
    print("enter valid integer number")
except ZeroDivisionError:
    print("zero not allowed")
```

# multiple exception handling by one handler

```python
try:
    x=int(input("enter the number = "))
    result=100/x
    print(result)
except (ValueError,ZeroDivisionError):  #it should give by tuple only.
    print("invalid input")
#used when multple handler requires same response 
```

```python
try:
    age=int(input("enter the age = "))
except ValueError:
    print("invalid age")
else:
    print("the age you entered is = ",age)

```

```python
#finally
try:
    print(10/0)
except ZeroDivisionError:
    print("zero not allowed")
finally:
    print("program completed")
#finally used for file cleanup
```

```python
#example
file =None 
try:
    file=open("dataset\data.txt","r")
    data=file.read()
except FileNotFoundError:
    print("file not found")
else:
    print(data)
finally:
    if file:
        file.close()
#if file opens it properly closed 
#however moder python used with for better file handling 
#with used context manager __enter__ and __exit__ methods 
```

# Exception type +message print

```python
try:
    result=10/0
except Exception as e:
    print("exception name = ",type(e).__name__)
    print("error message = ",e)
    
#this pattern very useful when logging unexpected failute
```

# bad coding 

```python
try:
    result=10/0
except Exception:
    pass
finally:
    print("program completed")
#donot use always except Exception. because it can hide programming bugs 
#above code works but we never discover what is actual bug 
```

```python
#above code. we didnot able to find what error produced but code completed successfully 
#donot write code like that.
#good way is see below code 
try:
    result=10/0
except Exception as e:
    print("error type = ",type(e).__name__)
    print("error message = ",e)
finally:
    print("program completed")
#we got exception and know how to handle 
```

# raise exceptions 

```python
#we can manually generate the exceptions 
age=-25
if age<0:
    raise ValueError("age cannot be negative number")
```

```python
age=-25

try:
    if age<0:
        raise ValueError("age cannot be negative number")
except Exception as e:
    print(e)
```

```python
#validation and raise error 

def set_age(age):
    if age<0:
        raise ValueError("age must be positive number")
    return age

age=-10
try:
    age=set_age(age)
except Exception as e:
    print(e)
else:
    print("the age is = ",age)
finally:
    print("program completed successfully")
```

```python
#raise without exception 
try:
    x=int("abc")
except Exception as e:
    print("logging error")
    raise

#here after raise no code because higher level decide what to do 
#just catch the error print the log and did some action and raise will do high level code will handle it 
#we donot worry here what to do.
```

# custom exception - advanced level

```python
#let create own exception 

class InvalidAgeError(Exception):
    pass 
age=-10
if age<0:
    raise InvalidAgeError("age cannot be negative")
```

```python
#custom exception with data 

class InsufficientBalanceError(Exception):
    def __init__(self,balance_amount,withdraw_amount):
        self.balance_amount=balance_amount
        self.withdraw_amount=withdraw_amount
        
        super().__init__(f'balance={balance_amount},required={withdraw_amount}')
    
#usage
balance_amount=1000
withdraw_amount=5000

if balance_amount<withdraw_amount:
    raise InsufficientBalanceError(balance_amount,withdraw_amount)
    
#custom error, custom data, custom error message 
```
