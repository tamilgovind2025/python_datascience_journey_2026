```python
error=[]
for data in dir(builtins):
    if data.endswith("Error"):
        error.append(data)
        
print(error)
print(len(error))
```

```python
try:
    age=int(input("enter the age = "))
    print(age)
except ValueError:
    print("enter valid number")
print("program completed successfully")
```

```python
try:
    age=int(input("enter the age = "))
    print(age)
except ValueError:
    print("enter valid number")
print("program completed successfully") #now our program cannot crash. it handled properly
```
