# multiple exception handling 

```python
try:
    x=int(input("enter the number "))
    result=100/x
    print(result)
except ValueError:
    print("enter valid integer number")
except ZeroDivisionError:
    print("zero not allowed")
```

# multiple exception handling by one handler

```python
try:
    x=int(input("enter the number = "))
    result=100/x
    print(result)
except (ValueError,ZeroDivisionError):  #it should give by tuple only.
    print("invalid input")
#used when multple handler requires same response 
```
