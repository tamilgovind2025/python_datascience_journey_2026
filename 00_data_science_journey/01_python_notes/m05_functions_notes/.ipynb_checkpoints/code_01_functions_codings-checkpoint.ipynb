```python
#becareful 

def add(a,b):
    print(a+b)
result=add(10,20)
print("the total value = ",result) #None will print because we have no return statement in funciton definition 
```

```python
#a function can return more values 
#ex 
def arithmatic_calculator(a,b):
    add=a+b
    sub=a-b
    mul=a*b
    div=a/b
    mod=a%b
    
    return add,sub,mul,div,mod #by default it returns tuple format 

add1,sub1,mul1,div1,mod1=arithmatic_calculator(10,20)
print("addition = ",add1)
print("subtraction = ",sub1)
print("multiplication = ",mul1)
print("division = ",div1)
print("modulus = ",mod1)

result=arithmatic_calculator(3,2)
print(result)
print(type(result))

print("addition value = ",result[0])
print("subtraction value = ",result[1])
```

## 5 positional arguments

## 9 arbitary arguments 


```python
def add(a,b):
    return a+b
add(10,20)
#but this code dosenot work 
add(10,20,30)
```

```python
#to overcome we use arbitary arguments 

def add(*data):
    print(data)
    print(type(data))
    return sum(data)

total= add(10,20)
print(total)

total= add(10,20,30)
print(total)

total= add(10,20,30,-10)
print(total)
```

## 10 arbitary keyword arguments 

```python
# arbitary keyword arguments 

def student(**kwargs):
    print(kwargs)
    print(type(kwargs))
    
student(name="govind",age=22,dept="ECE",year=2,college="Anna university")
student(name="govind")
```

```python
def employee(**data):
    print(type(data))
    for key,value in data.items():
        print(key,value)
        
employee(name="govind",dept="sales",salary=20_000,age=35)
```

```python
#10 combine arbitary arguments and arbitary keywords arguments 

def demo(*args,**kwargs):
    print(args)
    print(type(args))
    print(kwargs)
    print(type(kwargs))
    
demo(10,20,30,40,name="vinoth",age=25,dept="sales")
```

```python

def demo(**kwargs,*args): #this code will not work positional first come 
    print(args)
    print(type(args))
    print(kwargs)
    print(type(kwargs))
    
demo(name="vinoth",age=25,dept="sales",10,20,30,40,)
```

## global variable 


```python
x=200 

def test_func():
    print("global value of x = ",x) 
    #here read only. we cannot modifiy the global variable 

test_func()
print("value of x = ",x)
```

```python
#modify global variable inside function use keyword global 
#but try to avoid global varibles 
#it will exist to our entire project / applications 

x=200 

def test_func2():
    global x
    x=x+500
    print("the global x value = ",x)
test_func2()
print("x value = ",x)
```

## non local

```python
#non local used in nested funcntion 
def outer():
    data=100
    def inner():
        print("the data value = ",data) #here you can read the outer function data but cannot modify 
    inner()
outer()
```

```python
#non local used in nested funcntion 
def outer():
    data=100
    def inner():
        nonlocal data
        data=data+500
        print("the data value inside inner = ",data) #here you can read the outer function data but cannot modify 
    inner()
    print("the data value  outer function = ",data)
    
outer()
```

```python
#nested function 

def outer():
    print("i am entering outer function")
    def inner():
        print("i am entering inner function ")
    inner()
outer()
```

```python
#closure 
def outer():
    count=0
    def inner():
        nonlocal count 
        count+=1
        return count 
    return inner

counter=outer()
print(counter())
print(counter())
print(counter())
```

```python
def greet():
    print("hello")
x=greet

x()
```

```python
def outer():
    count=0
    def inner():
        nonlocal count
        count+=1
        return count 
    return inner
obj=outer()
print(obj())
print(obj())
print(obj())
```

```python
#lambda anonymous function or lambda functions 
#how to write general function 

def square(x):
    return x*x
print(square(5))

#lambda function
fun_a=lambda x:x*x
print(fun_a(5))
```

```python
#lambda with sorted 
students=[("govind",97),("kavitha",99),("vinoth",87),("gayu",92)]
students.sort(key=lambda x:x[1])
students
```

```python
#recursion 
#a function can call itselt called recursion 

def count_down(n):
    if n==0:
        return 
    print(n)
    count_down(n-1)
```

```python
count_down(5)
```
