```python
#Real-Time Function Example — Employee Salary

def calculate_salary(basic,hra,da,others,bonus):
    total=basic+hra+da+others+bonus
    return total 

net_salary=calculate_salary(12000,5000,3000,2000,0)
print("net salary = ",net_salary)
```

```python
#code 
import pandas as pd 

path=r"data_set\data_science_functions_employee_dataset.xlsx"

df=pd.read_excel(path)
df.head(2)
```

```python

#salary summary 

def salary_summary(df):
    
    data={
        "total":df["Salary"].sum(),
        "average":df["Salary"].mean(),
        "median":df["Salary"].median(),
        "max": df["Salary"].max(),
        "min": df["Salary"].min(),
    }
    
    return data

data=salary_summary(df)
print(data)
```

```python
#real time ml funcntion 

def calculate_accuracy(data_predicted,exact_value):
    if exact_value==0:
        return 0
    return data_predicted/exact_value*100

acc=calculate_accuracy(99,100)
print(acc)
acc=calculate_accuracy(25,28)
print(acc)
```
