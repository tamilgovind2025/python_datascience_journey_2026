```python
#type hints 
#professional coder uses type hints for better visualization 

def add(a:int,b:int)-> int:
    return a+b
```

```python
add(10,20)
print(add.__doc__)
```

```python
add(10.5,20.5)
```

```python
#doc strings 

def calculate_area(radius:float)->float:
    """
    calculate area of circle 
    
    args.
    radius: radius of circle 
    
    return.
    return area of circle 
    """
    return 3.14159*radius*radius

#print(calculate_area.__doc__)
help(calculate_area)
```

```python
print(calculate_area(3))
```

```python
#early return a function can return more than a time 

def check_age(age:int)->bool:
    if age>=18:
        return True
    return False
print(check_age(26))
print(check_age(6))
```

```python
help(check_age)
```

```python
#validate functions 

def validate_email(email:str)->bool:
    if "@" not in email:
        return False
    return True

print(validate_email("govind@gmail.com"))
print(validate_email("govind12345"))
```

```python
#exception handling in python functions 
def divide(a:int,b:int)->float:
    try:
        return a/b
    except Exception as e:
        return 
print(divide(10,0))
print(divide(10,20))
```

```python
#raise exception 
def withdraw(balance:float,amount:float):
    if amount<0:
        raise ValueError("withdraw amount cannnot be negative")
    if balance<amount:
        raise ValueError("insufficient funds in your account")
    return balance-amount
        
print("after withdrawn your balance is ",withdraw(5000,1000))
```

```python
#function composition 

def clean_name(name:str)->str:
    return name.strip().title()

def create_email(name):
    return clean_name(name).lower()+"@zenmatic.in"

print(create_email("        govind        "))
```
